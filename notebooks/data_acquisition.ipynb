```python
import pandas as pd
import glob
import os
```

```python
# Path 
file_paths = glob.glob("..//Transaction_data-23-25//shop-wise-trans-details_*_2023.csv") + \
             glob.glob("..//Transaction_data-23-25//shop-wise-trans-details_*_2024.csv") + \
             glob.glob("..//Transaction_data-23-25//shop-wise-trans-details_*_2025.csv")

print(f"Found {len(file_paths)} files")

dfs = []
for fp in file_paths:
    filename = os.path.basename(fp)
    year = filename.split("_")[-1].replace(".csv", "")

    # Read CSV and add year column
    df = pd.read_csv(fp)
    df['year'] = int(year)

    dfs.append(df)

# Combine all into one master DataFrame
master_df = pd.concat(dfs, ignore_index=True)
```

```python
# Step 2: Load and combine all Card Status CSVs (year-wise)
card_files = glob.glob("..//Shop_Card_Status-23-25//fpshop-card-status_*_2023.csv") + \
             glob.glob("..//Shop_Card_Status-23-25//fpshop-card-status_*_2024.csv") + \
             glob.glob("..//Shop_Card_Status-23-25//fpshop-card-status_*_2025.csv")

card_dfs = []
for fp in card_files:
    filename = os.path.basename(fp)
    year = filename.split("_")[-1].replace(".csv", "")
    df = pd.read_csv(fp)
    df['year'] = int(year)
    card_dfs.append(df)

card_status_df = pd.concat(card_dfs, ignore_index=True)
```

```python
# Step 3: Load FPS Locations 
fps_locations_df = pd.read_csv("..//data/shop-status-details_6_2025.csv")

print(master_df.shape)
print(card_status_df.shape)
print(fps_locations_df.shape)
```

```python
# Step 4: Triple Join on shopNo, distCode, and year
unified_df = master_df.merge(card_status_df, on=['shopNo','distCode','year'], how='inner') \
                            .merge(fps_locations_df, on=['shopNo','distCode'], how='inner')
print(unified_df.shape)
```

```python
# Step 5: Save unified dataset
unified_df.to_csv("..//data//unified_master.csv", index=False)
```
