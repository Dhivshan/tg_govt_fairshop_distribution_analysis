```python
# 1. Imports
import pandas as pd
import numpy as np

# Load unified dataset
df = pd.read_csv("C://Users//cxsha//PycharmProjects//Telangana_govt_ration_distribution_analysis//data//unified_master.csv")  # adjust path if needed
df.head()
```

```python
# --- Feature Engineering ---

# 1. Utilization Ratio (Transactions per Ration Card)
df["utilization_ratio"] = df["noOfTrans"] / (df["totalRcs"] + 1)

commodities = ['riceAfsc','riceFsc','riceAap','wheat']

df['rice_total'] = df['riceAfsc'] + df['riceFsc'] + df['riceAap']

# 2. Commodity Intensity (Rice-to-Wheat ratio)
df["rice_wheat_ratio"] = df["rice_total"] / (df["wheat"] + 1)

# 3. Volatility Features (transaction std dev per shop)
df["trans_volatility"] = df.groupby("shopNo")["noOfTrans"].transform("std")

# 5. Policy Impact Dummy (ONORC rollout after 2023)
df["onor_policy"] = np.where(df["year"] >= 2023, 1, 0)

# 7. Lag Features (previous month transactions)
df["lag_trans"] = df.groupby("shopNo")["noOfTrans"].shift(1)

# 8. Rolling Mean (3-month average transactions)
df["rolling_mean_3m"] = df.groupby("shopNo")["noOfTrans"].transform(lambda x: x.rolling(3).mean())

# --- Save Engineered Dataset ---
df.to_csv("C://Users//cxsha//PycharmProjects//Telangana_govt_ration_distribution_analysis//data//engineered_features.csv", index=False)

print("✅ Feature engineering complete. Saved to data/engineered_features.csv")
```
